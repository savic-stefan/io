### Занятие 2. Экстремум функции нескольких переменных. Метод множителей Лагранжа
1. Используя метод множителей Лагранжа найти минимум функции: 𝑓(𝑥,𝑦)=𝑥2+𝑦2−4𝑥−2𝑦+5=0,
при ограничениях 𝑥+𝑦=3.

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Определение символьных переменных
x, y, lambda_ = sp.symbols('x y lambda')

# Определение функций
f = x**2 + y**2 - 4*x - 2*y + 5
g = x + y - 3

# Функция Лагранжа
L = f - lambda_ * g

# Частные производные
dL_dx = sp.diff(L, x)
dL_dy = sp.diff(L, y)
dL_dlambda = sp.diff(L, lambda_)

# Решение системы уравнений
eq1 = sp.Eq(dL_dx, 0)
eq2 = sp.Eq(dL_dy, 0)
eq3 = sp.Eq(dL_dlambda, 0)
solution = sp.solve((eq1, eq2, eq3))
print(f"Минимум функции достигается в точке ({solution[x]}, {solution[y]}) и равен {solution[lambda_]}. ")
```

Используя метод множителей Лагранжа найти минимум функции: 𝑓(𝑥,𝑦)=3𝑥2+4𝑦2−6𝑥−8𝑦+10=0,
при ограничениях 𝑥+𝑦=5.

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Определение символьных переменных
X, Y, Lambda_ = sp.symbols('x y lambda')

F = 3*X**2 + 4*Y**2 - 6*X - 8*Y + 10
G = X + Y - 5

# Функция Лагранжа
L1 = F - Lambda_ * G

# Частные производные
dL_dx1 = sp.diff(L1, X)
dL_dy1 = sp.diff(L1, Y)
dL_dlambda1 = sp.diff(L1, Lambda_)

# Решение системы уравнений
Eq1 = sp.Eq(dL_dx1, 0)
Eq2 = sp.Eq(dL_dy1, 0)
Eq3 = sp.Eq(dL_dlambda1, 0)
Solution = sp.solve((Eq1, Eq2, Eq3))
print(f"Минимум функции достигается в точке ({Solution[X]}, {Solution[Y]}) и равен {Solution[Lambda_]}. ")
```
