### Производственные функции. Экстремум функции нескольких переменных.

1. Дана производственная функция вида: 𝑄(𝑥,𝑦)=2𝑥2+3𝑥𝑦+4𝑦2−6𝑥−12𝑦+10,
𝑥,𝑦 – переменные, определяющие количество использованных ресурсов.
Найти экстремум функции и параметры, при которых он достигается.

```python
from sympy import symbols, diff, solve, Eq, hessian, Matrix

def find_extremum(Q, x, y):
    dQ_dx = diff(Q, x)
    dQ_dy = diff(Q, y)
    
    critP = solve((Eq(dQ_dx, 0), Eq(dQ_dy, 0)), (x, y))
    H = hessian(Q, (x, y))
    
    if critP:
        x_val = critP[x]
        y_val = critP[y]
        det_H = H.det().subs({x: x_val, y: y_val})
        d2Q_dx2 = H[0, 0].subs({x: x_val, y: y_val})
        
        if det_H > 0 and d2Q_dx2 > 0:
            extremum_type = "минимум"
        elif det_H > 0 and d2Q_dx2 < 0:
            extremum_type = "максимум"
        else:
            extremum_type = "седловая точка"
        
        Q_value = Q.subs({x: x_val, y: y_val})
        
        return {
            "Критическая точка": (x_val, y_val),
            "Тип экстремума": extremum_type,
            "Значение функции": Q_value,
            "матрица Гессе": H
        }
    else:
        return None
    
x, y = symbols('x y')
Q1 = 2*x**2 + 3*x*y + 4*y**2 - 6*x - 12*y + 10
res1 = find_extremum(Q1, x, y)
print('Задача 1 \n',res1)
```

2. Дана производственная функция вида: 𝑄(𝑥,𝑦)=𝑥2+𝑦2−4𝑥−2𝑦+5,
𝑥,𝑦 – переменные, определяющие количество использованных ресурсов.
Найти экстремум функции и параметры, при которых он достигается.

```python
Q2 = x**2 + y**2 - 4*x - 2*y + 5
res2 = find_extremum(Q2, x, y)
print('Задача 2 \n',res2)
```
